```r
# Paper criteria: |log2FC| > 1  AND  adj.P.Val < 0.05
apply_threshold <- function(df, label) {

  df$sig <- "Not significant"
  df$sig[df$adj.P.Val < 0.05 & df$logFC >  1] <- "Up-regulated"
  df$sig[df$adj.P.Val < 0.05 & df$logFC < -1] <- "Down-regulated"

  # Convert to ordered factor for consistent plot ordering
  df$sig <- factor(df$sig,
                   levels = c("Up-regulated", "Down-regulated", "Not significant"))

  # Summary
  cat("\n", label, "\n")
  cat("  Total DEGs        :", nrow(df), "\n")
  cat("  Up-regulated      :", sum(df$sig == "Up-regulated"), "\n")
  cat("  Down-regulated    :", sum(df$sig == "Down-regulated"), "\n")
  cat("  Not significant   :", sum(df$sig == "Not significant"), "\n")

  # Top 5 by logFC
  cat("\n  Top 5 by logFC:\n")
  top5 <- df %>%
    arrange(desc(logFC)) %>%
    select(id, logFC, adj.P.Val) %>%
    head(5)
  print(top5, row.names = FALSE)

  return(df)
}

ra <- apply_threshold(ra, "Rheumatoid Arthritis")
uc <- apply_threshold(uc, "Ulcerative Colitis")
```

```r
# Define hub genes from paper for annotation
hub_genes      <- c("CXCL13","CCR7","CCL19","CD19","GZMB",
                    "CXCL9","CD38","CXCR4","IL7R","SELL")
validated_hubs <- c("CCR7","CD19","CXCL13","CXCR4","SELL")

# Add -log10(adj.P) column to both
add_neglogp <- function(df) {
  df %>%
    mutate(
      neg_log10P  = -log10(pmax(adj.P.Val, 1e-300)),
      is_hub      = id %in% hub_genes,
      is_validated = id %in% validated_hubs,
      label       = ifelse(is_hub, id, NA_character_)
    )
}

ra2 <- add_neglogp(ra)
uc2 <- add_neglogp(uc)

# Significance threshold line
p_threshold <- -log10(0.05)

# Function to build one volcano panel
make_volcano <- function(df, title) {
  ggplot(df, aes(x = logFC, y = neg_log10P, colour = sig)) +

    # Background (not significant) points first
    geom_point(data = filter(df, sig == "Not significant"),
               alpha = 0.25, size = 1.2) +

    # Significant points on top
    geom_point(data = filter(df, sig != "Not significant"),
               alpha = 0.65, size = 2) +

    # Threshold lines
    geom_vline(xintercept =  1, linetype = "dashed",
               colour = "grey50", linewidth = 0.7) +
    geom_vline(xintercept = -1, linetype = "dashed",
               colour = "grey50", linewidth = 0.7) +
    geom_hline(yintercept = p_threshold, linetype = "dotted",
               colour = "grey50", linewidth = 0.7) +

    # Hub gene labels — ggrepel avoids overlap
    geom_label_repel(
      data        = filter(df, is_hub),
      aes(label   = label),
      size        = 2.8,
      fontface    = "bold",
      colour      = ifelse(filter(df, is_hub)$is_validated,
                           "#E63946", "#023E8A"),
      fill        = "white",
      box.padding = 0.4,
      max.overlaps = 20,
      segment.colour = "grey60",
      segment.size   = 0.4
    ) +

    scale_colour_manual(
      values = c(
        "Up-regulated"    = "#E63946",
        "Down-regulated"  = "#457B9D",
        "Not significant" = "#ADB5BD"
      ),
      labels = c(
        "Up-regulated"    = paste0("Up (n=", sum(df$sig=="Up-regulated"), ")"),
        "Down-regulated"  = paste0("Down (n=", sum(df$sig=="Down-regulated"), ")"),
        "Not significant" = "Not significant"
      )
    ) +

    labs(
      title    = title,
      x        = "log_2 Fold Change",
      y        = "-log_10(adj. P-value)",
      colour   = NULL
    ) +
    annotate("text",
             x    = max(df$logFC) * 0.85,
             y    = p_threshold + 0.3,
             label = "adj.P = 0.05",
             size  = 2.8,
             colour = "grey40") +
    theme_classic(base_size = 11) +
    theme(
      plot.title      = element_text(face = "bold", size = 11),
      legend.position = c(0.15, 0.85),
      legend.background = element_rect(fill = "white", colour = "grey80")
    )
}

v_ra <- make_volcano(ra2, "Rheumatoid Arthritis")
v_uc <- make_volcano(uc2, "Ulcerative Colitis")

# Combine side by side using patchwork
p3 <- (v_ra | v_uc) +
  plot_annotation(
    title    = "Obj 1 — Statistical Significance vs Fold Change",
    subtitle = "Red labels = validated hub genes | Blue labels = other hub genes",
    theme    = theme(
      plot.title    = element_text(face = "bold", size = 13),
      plot.subtitle = element_text(size = 10, colour = "grey40")
    )
  )

print(p3)
ggsave("obj1_significance_scatter.png", p3,
       width = 14, height = 6, dpi = 150)
```

```r
# ── Cleaner version of prep_score ─────────────────────────────────────────
prep_score <- function(df, top_n = 10) {

  df <- df %>%
    mutate(
      negP      = -log10(pmax(adj.p.val, 1e-300)),
      score     = sqrt(logfc^2 + negP^2),
      candidate = adj.p.val < 0.05 & abs(logfc) > 1,
      color     = "Other",
      label     = NA_character_
    )

  # Top N up-regulated — ranked by score within up genes only
  top_up <- df %>%
    filter(candidate & logfc > 1) %>%
    arrange(desc(score)) %>%
    slice_head(n = top_n) %>%
    pull(gene)

  # Top N down-regulated — ranked by score within down genes only
  top_down <- df %>%
    filter(candidate & logfc < -1) %>%
    arrange(desc(score)) %>%
    slice_head(n = top_n) %>%
    pull(gene)

  df %>%
    mutate(
      color = case_when(
        gene %in% top_up   ~ "Top Up",
        gene %in% top_down ~ "Top Down",
        TRUE               ~ "Other"
      ),
      label = ifelse(color != "Other", gene, NA)
    )
}

ra2 <- prep_score(ra, top_n = 10)
uc2 <- prep_score(uc, top_n = 10)

# Quick check
cat("RA  — Top Up:", sum(ra2$color == "Top Up"),
    "| Top Down:", sum(ra2$color == "Top Down"), "\n")
cat("UC  — Top Up:", sum(uc2$color == "Top Up"),
    "| Top Down:", sum(uc2$color == "Top Down"), "\n")

# ── Plot function ──────────────────────────────────────────────────────────
make_plot <- function(df, title) {

  df_other <- filter(df, color == "Other")
  df_top   <- filter(df, color != "Other")

  ggplot(df, aes(x = logfc, y = negP)) +

    # Grey background — all non-top genes
    geom_point(data  = df_other,
               color = "#CCCCCC",
               size  = 0.8,
               alpha = 0.5) +

    # Top genes — coloured and larger
    geom_point(data = df_top,
               aes(color = color),
               size  = 3.5,
               alpha = 0.9) +

    # Labels on top genes only
    geom_text_repel(
      data          = df_top,
      aes(label     = label),
      size          = 3,
      fontface      = "bold",
      box.padding   = 0.5,
      max.overlaps  = Inf,
      seed          = 42,
      segment.size  = 0.3,
      segment.color = "grey40",
      min.segment.length = 0
    ) +

    # Threshold lines
    geom_vline(xintercept = c(-1, 1),
               linetype  = "dashed",
               color     = "grey30",
               linewidth = 0.5) +
    geom_hline(yintercept = -log10(0.05),
               linetype  = "dotted",
               color     = "grey30",
               linewidth = 0.5) +

    scale_color_manual(
      values = c(
        "Top Up"   = "#E63946",
        "Top Down" = "#457B9D"
      ),
      labels = c(
        "Top Up"   = paste0("Top Up"),
        "Top Down" = paste0("Top Down")
      )
    ) +

    labs(
      title = title,
      x     = expression(log[2]~"Fold Change"),
      y     = expression(-log[10]~"(adj. P-value)"),
      color = NULL
    ) +

    theme_classic(base_size = 12) +
    theme(
      plot.title      = element_text(face = "bold", hjust = 0.5, size = 12),
      legend.position = "bottom",
      legend.text     = element_text(size = 10)
    )
}

# ── Build and combine ──────────────────────────────────────────────────────
p_ra <- make_plot(ra2, "Rheumatoid Arthritis")
p_uc <- make_plot(uc2, "Ulcerative Colitis")

p_final <- (p_ra | p_uc) +
  plot_annotation(
    title    = "Top 10 DEGs - RA vs UC",
    theme    = theme(
      plot.title    = element_text(face = "bold", size = 14, hjust = 0.5),
      plot.subtitle = element_text(size = 9,  hjust = 0.5, colour = "grey40")
    )
  )

print(p_final)

# ── Ranked tables ──────────────────────────────────────────────────────────
cat("\nTop RA genes:\n")
ra2 %>%
  filter(color != "Other") %>%
  arrange(color, desc(score)) %>%
  select(gene, logfc, adj.p.val, score, color) %>%
  print()

cat("\nTop UC genes:\n")
uc2 %>%
  filter(color != "Other") %>%
  arrange(color, desc(score)) %>%
  select(gene, logfc, adj.p.val, score, color) %>%
  print()

cat("\nTop genes common to BOTH diseases:\n")
common_top <- intersect(
  filter(ra2, color != "Other")$gene,
  filter(uc2, color != "Other")$gene
)
print(common_top)
```

```r
library(tidyverse)

# -------------------------
# 1. Clean column names
# -------------------------
colnames(ra) <- tolower(colnames(ra))
colnames(uc) <- tolower(colnames(uc))

names(ra)[1] <- "gene"
names(uc)[1] <- "gene"


# -------------------------
# 3. Filter significant DEGs
# -------------------------
ra_sig <- ra %>%
  filter(adj.p.val < 0.05 & abs(logfc) > 1)

uc_sig <- uc %>%
  filter(adj.p.val < 0.05 & abs(logfc) > 1)

# -------------------------
# 4. Summary counts
# -------------------------
cat("RA significant DEGs:", nrow(ra_sig), "\n")
cat("UC significant DEGs:", nrow(uc_sig), "\n")

# -------------------------
# 5. Shared DEGs
# -------------------------
shared_genes <- intersect(ra_sig$gene, uc_sig$gene)
cat("Shared DEGs:", length(shared_genes), "\n")

# -------------------------
# 6. Merge shared genes for comparison
# -------------------------
merged <- inner_join(
  ra_sig %>% dplyr::select(gene, logfc),
  uc_sig %>% dplyr::select(gene, logfc),
  by = "gene",
  suffix = c("_ra", "_uc")
)

# -------------------------
# 7. Direction consistency
# -------------------------
merged <- merged %>%
  mutate(
    same_direction = sign(logfc_ra) == sign(logfc_uc)
  )

cat("Directional agreement table:\n")
print(table(merged$same_direction))

# -------------------------
# 8. Percentage consistency
# -------------------------
consistency_rate <- mean(merged$same_direction) * 100
cat("Directional consistency (%):", consistency_rate, "\n")

# -------------------------
# 9. Top shared genes
# -------------------------
cat("\nTop shared genes (by RA effect size):\n")
print(
  merged %>%
    arrange(desc(abs(logfc_ra))) %>%
    head(10)
)

# -------------------------
# 10. Correlation of effect sizes
# -------------------------
correlation <- cor(merged$logfc_ra, merged$logfc_uc, method = "spearman", use = "complete.obs")
cat("\nSpearman correlation:", correlation, "\n")
```

```r
library(dplyr)
library(ggplot2)

# -------------------------
# Base top genes
# -------------------------
base_top <- ra %>%
  filter(adj.p.val < 0.05 & abs(logfc) > 1) %>%
  arrange(adj.p.val) %>%
  head(10)

# -------------------------
# Store results
# -------------------------
results <- data.frame(i = numeric(), consistency = numeric())

# -------------------------
# Loop thresholds
# -------------------------
for(i in seq(0.1, 5, by = 0.01)) {

  relaxed <- ra %>%
    filter(adj.p.val < 0.001 & abs(logfc) > i) %>%
    arrange(adj.p.val) %>%
    head(10)

  consistency <- length(intersect(base_top$gene, relaxed$gene)) /
    length(base_top$gene) * 100

  results <- rbind(results, data.frame(i = i, consistency = consistency))
}

# -------------------------
# Find first drop point (100 → <100)
# -------------------------
drop_point <- results %>%
  filter(consistency < 100) %>%
  slice(1) %>%
  pull(i)

# -------------------------
# LINE PLOT
# -------------------------
ggplot(results, aes(x = i, y = consistency)) +
  geom_line(color = "steelblue", size = 1) +
  geom_point(size = 1.5, color = "steelblue") +

  # Red vertical line at first drop
  geom_vline(xintercept = drop_point, color = "red", linetype = "dashed", size = 1) +

  theme_minimal() +
  labs(
    title = "Stability of Top DEGs under Parameter Changes",
    x = "logFC threshold (i)",
    y = "Consistency with baseline (%)"
  )
```
