```r
# Install packages
if (!require("BiocManager", quietly = TRUE))
    install.packages("BiocManager")
BiocManager::install()

BiocManager::install("DESeq2")
BiocManager::install("AnnotationDbi")
BiocManager::install("org.Hs.eg.db")

library(DESeq2)
library(AnnotationDbi)
library(org.Hs.eg.db)
library(ggplot2)
library(dplyr)
```

```r
# Read in data
ra_deg <- read.csv("/content/ra_deg.csv")

# MA plot
with(ra_deg, plot(
  AveExpr,
  logFC,
  pch = 16,
  cex = 0.5,
  col = ifelse(adj.P.Val < 0.05, "red", "grey"),
  xlab = "Mean Expression",
  ylab = "Log2 Fold Change",
  main = "MA Plot"
))
```

```r
# Volcano plot

# Set plotting parameters
par(mar = c(5,5,5,5), cex = 1.0, cex.main = 1.4, cex.axis = 1.4, cex.lab = 1.4)

# Pick top 10 by smallest padj (highest -log10(padj))
ra_sig <- head(ra_deg[order(ra_deg$adj.P.Val), ], 10)

# Make plot
with(ra_deg, plot(
  logFC,
  -log10(adj.P.Val),
  pch = 19,
  col = "#c5cfa0ff",
  xlab = expression(bold(~Log[2]~Fold~Change)),
  ylab = expression(bold(~-Log[10]~Adjusted~P~Value)),
  ylim = c(0, 215)
))
```
