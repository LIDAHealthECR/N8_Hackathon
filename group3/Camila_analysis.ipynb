```r
dds <- DESeqDataSetFromMatrix(
  countData = countData,
  colData = colData,
  design = ~ group
)
```

```r
dds <- DESeq(dds)
```

```r
res <- results(dds)
```

```r
head(res)
```

```r
# =========================
# VOLCANO PLOT
# Differentially Expressed Genes
# =========================

# Load required libraries
library(ggplot2)
library(ggrepel)

# Convert DESeq2 results into a dataframe
res_df <- as.data.frame(res)

# Remove missing values
res_df <- na.omit(res_df)

# Add gene names as a new column
res_df$gene <- rownames(res_df)

# Create significance categories
res_df$group <- "Not Significant"

# Upregulated genes
res_df$group[
  res_df$padj < 0.05 &
  res_df$log2FoldChange > 1
] <- "Upregulated"

# Downregulated genes
res_df$group[
  res_df$padj < 0.05 &
  res_df$log2FoldChange < -1
] <- "Downregulated"

# Select top significant genes for labeling
top_genes <- res_df[
  order(res_df$padj),
][1:15, ]

# Generate volcano plot
ggplot(
  res_df,
  aes(
    x = log2FoldChange,
    y = -log10(padj),
    color = group
  )
) +

# Plot points
geom_point(
  alpha = 0.7,
  size = 2
) +

# Define colors
scale_color_manual(
  values = c(
    "Upregulated" = "red",
    "Downregulated" = "blue",
    "Not Significant" = "gray"
  )
) +

# Add threshold lines
geom_vline(
  xintercept = c(-1, 1),
  linetype = "dashed"
) +

geom_hline(
  yintercept = -log10(0.05),
  linetype = "dashed"
) +

# Add gene labels
geom_text_repel(
  data = top_genes,
  aes(label = gene),
  size = 3,
  max.overlaps = 20
) +

# Apply theme
theme_classic() +

# Add plot labels
labs(
  title = "Volcano Plot of Differentially Expressed Genes",
  subtitle = "Comparison between Disease and Control groups",
  x = "log2 Fold Change",
  y = expression(-log[10]("Adjusted p-value")),
  color = "Gene Category"
)
```

```r
# =========================
# LIBRARIES
# =========================

library(DESeq2)
library(pheatmap)
library(RColorBrewer)

# =========================
# VARIANCE STABILIZATION
# =========================

vsd <- vst(dds)

# =========================
# SELECT TOP GENES
# =========================

topgenes <- head(order(res$padj), 100)

# expression matrix
mat <- assay(vsd)[topgenes, ]

# =========================
# SAMPLE ANNOTATION
# =========================

annotation <- data.frame(
  Type = colData$group
)

rownames(annotation) <- colnames(mat)

# =========================
# ANNOTATION COLORS
# =========================

ann_colors <- list(

  Type = c(
    "Control" = "#BDBD00",
    "Disease" = "#26C6DA"
  )

)

# =========================
# HEATMAP
# =========================

pheatmap(

  mat,

  # normalize each gene
  scale = "row",

  # clustering
  cluster_rows = TRUE,
  cluster_cols = TRUE,

  # annotations
  annotation_col = annotation,
  annotation_colors = ann_colors,

  # heatmap colors
  color = colorRampPalette(
    c("blue", "white", "red")
  )(100),

  # symmetric scale
  breaks = seq(-3, 3, length.out = 101),

  # labels
  show_colnames = FALSE,
  show_rownames = TRUE,

  # font sizes
  fontsize_row = 5,
  fontsize_col = 5,

  # borders
  border_color = "grey85",

  # dendrogram sizes
  treeheight_row = 60,
  treeheight_col = 60

)
```
